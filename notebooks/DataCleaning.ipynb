Phase 1: Data Preparation & Cleaning

Objective
The goal of this notebook is to ingest the raw IBM HR Analytics dataset and prepare it for exploratory data analysis (EDA) and machine learning.

Methodology
Data cleaning is a critical first step to ensure model accuracy and prevent 'garbage in, garbage out'. My workflow follows these structured steps:
1. Data Ingestion: Load the raw data to assess its initial shape and structure.
2. Integrity Checks: Identify and handle any missing values (NaNs) or duplicated records that could skew our analysis.
3. Feature Selection (Noise Reduction):** Remove columns that offer zero predictive value (e.g., constant values or unique identifiers) to reduce dimensionality and improve model efficiency.
4. Standardisation: Format column names for easier programmatic access in subsequent notebooks.

```python
# Import libraries
import pandas as pd
```

```python
# Load the dataset
df = pd.read_csv('Employee-Attrition-and-Performance.csv')
```

```python
# Integrity Checks (Proving the data is clean before manipulation)
print("--- Data Integrity Check ---")
print(f"Total Missing Values: {df.isnull().sum().sum()}")
print(f"Total Duplicated Rows: {df.duplicated().sum()}")
print("-" * 28)
```

```python
# RATIONALE FOR DROPPING COLUMNS (LO5 Fix):
# - 'EmployeeCount', 'Over18', and 'StandardHours' contain identical values for every row (zero variance). A model cannot learn from them.
# - 'EmployeeNumber' is a random unique ID. Leaving it in risks the model learning false correlations.
# Action: Drop these garbage columns to reduce noise and improve model efficiency.
df_clean = df.drop(columns=['EmployeeCount', 'Over18', 'StandardHours', 'EmployeeNumber'], errors='ignore')
```

```python
# Rename columns to lowercase (Best practice)
df_clean.columns = [col.lower() for col in df_clean.columns]
```

```python
# Save the clean file
df_clean.to_csv('clean_employee_data.csv', index=False)

print("Clean file created.")
print("Original Shape:", df.shape)
print("New Shape:", df_clean.shape)
```
